Load the 'paraphrase-MiniLM-L6-v2' model  from HuggingFace resource  using the  SentenceTransformer( *model-name* )  and store the reference to the model object in the 'model' variable

```python
model = SentenceTransformer("sentence-transformers/paraphrase-MiniLM-L6-v2")
print("Model loaded")
```

After loading the model, call the 'encode()' method on the model object to create a vector representation of a specific text sentence. Use your own text string  as the parameter.

```python
sentence = "Machine learning helps computers recognize patterns in data."

embedding = model.encode(sentence)

print("Sentence:", sentence)
print("Embedding shape:", embedding.shape)
print("First 10 values:", embedding[:10])
```

Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python
sentences_list = [
    "A university student uses a budgeting app to track daily expenses and save enough money for a summer trip with friends.",
    "A mobile finance application helps students monitor spending habits and plan savings for their next holiday without missing important monthly bills.",
    "The bank uses historical transaction patterns to identify unusual card payments and alert customers before a possible fraud attempt causes losses.",
    "A fraud detection system compares new purchases with previous banking activity to flag suspicious transactions and protect account holders from theft.",
    "After training at the gym, the athlete prepares a balanced meal with chicken, rice, and vegetables to support muscle recovery.",
    "A volleyball player plans meals with plenty of protein and carbohydrates because intense practice requires energy and adequate recovery time afterward.",
    "The weather forecast predicts heavy rain in the afternoon, so the hikers decide to postpone their trip to the mountain trail.",
    "A group of travelers checks the forecast before leaving because storms can make outdoor paths dangerous and spoil the planned hiking experience.",
    "A teacher organizes online lessons with short quizzes to help students practice new concepts and identify which topics need more explanation.",
    "An educational platform tracks quiz mistakes and recommends additional exercises so learners can improve their understanding of difficult technical subjects over time."
]

embeddings = model.encode(sentences_list)

print("Number of sentences:", len(sentences_list))
print("Embedding shape:", embeddings.shape)
print("Word counts:", [len(s.split()) for s in sentences_list])
```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
def cos_similarity_calculation(query, sentences, sentence_embeddings):
    query_embedding = model.encode(query)
    scores = util.cos_sim(query_embedding, sentence_embeddings)[0]

    results = sorted(
        zip(sentences, scores.tolist()),
        key=lambda x: x[1],
        reverse=True
    )

    return results

query = "How can a bank detect suspicious credit card activity?"

results = cos_similarity_calculation(query, sentences_list, embeddings)

for text, score in results[:3]:
    print(round(score, 4), text)
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def cosine_similarity_numpy(vector, batch):
    numerator = batch @ vector
    denominator = np.linalg.norm(batch, axis=1) * np.linalg.norm(vector)
    return numerator / (denominator + 1e-12)

query_embedding = model.encode(query)
scores = cosine_similarity_numpy(query_embedding, embeddings)

top_indices = np.argsort(scores)[::-1][:3]

for i in top_indices:
    print(round(float(scores[i]), 4), sentences_list[i])
```
